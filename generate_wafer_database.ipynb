```python
import pandas as pd
import numpy as np
import random

# ============================================================
# Semiconductor Wafer Yield Database Generator
# ============================================================

# -----------------------------
# CONFIGURATION
# -----------------------------
NUM_LOTS = 10
WAFERS_PER_LOT = 25          # 10 lots x 25 wafers = 250 rows
TOTAL_DIES = 1200

OUTPUT_FILE = "Generated_Wafer_Database.xlsx"

random.seed(42)
np.random.seed(42)

# ============================================================
# Process Steps
# ============================================================

PROCESS_STEPS = [
    "RCA Cleaning",
    "Thermal Oxidation",
    "Photolithography",
    "Etching",
    "Ion Implantation",
    "Annealing",
    "Metallization",
    "Passivation"
]

# ============================================================
# Lot Level Behaviour
# Every lot has its own process characteristics
# ============================================================

LOT_PROFILE = {
    1: {"particle_shift": 0, "oxide_shift": 0.00, "overlay_shift": 0, "implant_shift": 0, "metal_shift": 0},
    2: {"particle_shift": 2, "oxide_shift": 0.02, "overlay_shift": 0.3, "implant_shift": 0.1, "metal_shift": 0},
    3: {"particle_shift": 5, "oxide_shift": 0.03, "overlay_shift": 1.8, "implant_shift": 0.2, "metal_shift": 1},
    4: {"particle_shift": 1, "oxide_shift": -0.02, "overlay_shift": 0.5, "implant_shift": 0.3, "metal_shift": 0},
    5: {"particle_shift": 4, "oxide_shift": 0.05, "overlay_shift": 1.2, "implant_shift": 0.4, "metal_shift": 1},
    6: {"particle_shift": 9, "oxide_shift": 0.06, "overlay_shift": 2.0, "implant_shift": 0.6, "metal_shift": 2},
    7: {"particle_shift": 3, "oxide_shift": 0.01, "overlay_shift": 0.8, "implant_shift": 0.2, "metal_shift": 1},
    8: {"particle_shift": 7, "oxide_shift": 0.08, "overlay_shift": 2.5, "implant_shift": 0.8, "metal_shift": 3},
    9: {"particle_shift": 10, "oxide_shift": 0.10, "overlay_shift": 3.0, "implant_shift": 1.0, "metal_shift": 4},
    10: {"particle_shift": 2, "oxide_shift": 0.00, "overlay_shift": 0.4, "implant_shift": 0.2, "metal_shift": 0},
}

# ============================================================
# Root Cause Dictionary
# ============================================================

ROOT_CAUSE = {
    "Particle": "Particle Contamination",
    "Oxide Variation": "Oxide Thickness Variation",
    "Overlay Error": "Photolithography Overlay Drift",
    "Implant Error": "Implant Dose Drift",
    "Metal Void": "Metal Void Formation",
}

# ============================================================
# Generate Data
# ============================================================

rows = []
wafer_counter = 1

for lot in range(1, NUM_LOTS + 1):
    profile = LOT_PROFILE[lot]

    for w in range(WAFERS_PER_LOT):
        lot_id = f"L{lot:02d}"
        wafer_id = f"W{wafer_counter:03d}"
        wafer_counter += 1

        process = random.choice(PROCESS_STEPS)

        # ---------------------------------------
        # Generate Process Parameters
        # ---------------------------------------

        particle = max(0, int(np.random.normal(5 + profile["particle_shift"], 3)))

        oxide = round(np.random.normal(5.0 + profile["oxide_shift"], 0.04), 2)

        overlay = round(
            max(0.5, np.random.normal(2.5 + profile["overlay_shift"], 0.8)), 2
        )

        implant = round(
            max(0, np.random.normal(0.4 + profile["implant_shift"], 0.25)), 2
        )

        metal = max(0, int(np.random.poisson(1 + profile["metal_shift"])))

        # ---------------------------------------
        # Improved Yield Model
        # ---------------------------------------

        yield_loss = 0
        yield_loss += particle * 0.08                 # Particle contamination
        yield_loss += abs(oxide - 5.0) * 20            # Oxide thickness variation
        yield_loss += overlay * 0.25                    # Overlay error
        yield_loss += implant * 1.2                     # Implant dose variation
        yield_loss += metal * 0.8                        # Metal voids

        yield_percent = round(max(92, min(99.8, 100 - yield_loss)), 2)

        good_dies = round(TOTAL_DIES * yield_percent / 100)
        bad_dies = TOTAL_DIES - good_dies

        # ---------------------------------------
        # Major Defect
        # ---------------------------------------

        scores = {
            "Particle": particle,
            "Oxide Variation": abs(oxide - 5.0) * 20,
            "Overlay Error": overlay,
            "Implant Error": implant * 5,
            "Metal Void": metal * 2,
        }

        major = max(scores, key=scores.get)
        root = ROOT_CAUSE[major]

        # ---------------------------------------
        # Status
        # ---------------------------------------

        if yield_percent >= 98:
            status = "Pass"
        elif yield_percent >= 95:
            status = "Warning"
        else:
            status = "Fail"

        rows.append({
            "Lot_Id": lot_id,
            "Wafer_Id": wafer_id,
            "Process_Step": process,
            "Particle_Count": particle,
            "Oxide_Thickness_nm": oxide,
            "Overlay_Error_nm": overlay,
            "Implant_Dose_Error_%": implant,
            "Metal_Voids": metal,
            "Total_Dies": TOTAL_DIES,
            "Good_Dies": good_dies,
            "Bad_Dies": bad_dies,
            "Major_Defect": major,
            "Process_Status": status,
            "Yield_%": yield_percent,
            "Dominant_Root_Cause": root,
        })

# ============================================================
# Save Database
# ============================================================

df = pd.DataFrame(rows)
df.to_excel(OUTPUT_FILE, index=False)

# ============================================================
# Summary
# ============================================================

print("=" * 60)
print("SEMICONDUCTOR WAFER DATABASE GENERATED")
print("=" * 60)
print()
print("Total Lots :", NUM_LOTS)
print("Total Wafers :", len(df))
print("Average Yield :", round(df["Yield_%"].mean(), 2), "%")
print()
print(df["Process_Status"].value_counts())
print()
print("Saved As :", OUTPUT_FILE)
print("=" * 60)
```
